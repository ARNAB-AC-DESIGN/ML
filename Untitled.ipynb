```python
import cv2
from PIL import Image
import numpy as np
import os

def upscale_to_large_jpeg(input_path, output_path, scale=2, interpolation='lanczos', bg_color=(255,255,255), target_size_mb=10):
    img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)
    if img is None:
        raise FileNotFoundError(f"Cannot open image: {input_path}")

    h, w = img.shape[:2]

    interp_map = {
        'nearest': cv2.INTER_NEAREST,
        'bilinear': cv2.INTER_LINEAR,
        'bicubic': cv2.INTER_CUBIC,
        'lanczos': cv2.INTER_LANCZOS4
    }
    interp = interp_map.get(interpolation.lower(), cv2.INTER_LANCZOS4)

    current_scale = scale
    while True:
        upscaled = cv2.resize(img, (int(w * current_scale), int(h * current_scale)), interpolation=interp)

        # Handle alpha channel
        if upscaled.shape[2] == 4:
            alpha = upscaled[:, :, 3] / 255.0
            background = np.ones_like(upscaled[:, :, :3], dtype=np.uint8) * np.array(bg_color, dtype=np.uint8)
            upscaled_rgb = (upscaled[:, :, :3] * alpha[..., None] + background * (1 - alpha[..., None])).astype(np.uint8)
        else:
            upscaled_rgb = upscaled

        # Save temporarily with max quality
        upscaled_pil = Image.fromarray(cv2.cvtColor(upscaled_rgb, cv2.COLOR_BGR2RGB))
        upscaled_pil.save(output_path, format='JPEG', quality=100)

        # Check file size
        file_size_mb = os.path.getsize(output_path) / (1024 * 1024)
        print(f"Current scale: {current_scale}× → File size: {file_size_mb:.2f} MB")

        if file_size_mb >= target_size_mb:
            break
        else:
            current_scale += 0.5  # increase scale and try again

    print(f"✅ Upscaled JPEG saved: {output_path} ({current_scale}×, {file_size_mb:.2f} MB)")
    return output_path, current_scale

if __name__ == "__main__":
    print("=== PNG to Large JPEG Upscaler ===")
    input_file = input("Enter path to input PNG file: ").strip()
    output_file = input("Enter path for output JPEG file: ").strip()
    
    scale = float(input("Enter initial upscale factor (default 2): ") or 2)
    interpolation = input("Enter interpolation method [nearest/bilinear/bicubic/lanczos] (default lanczos): ") or 'lanczos'
    bg_color_input = input("Enter background color for transparency as R,G,B (default 255,255,255): ") or '255,255,255'
    bg_color = tuple(map(int, bg_color_input.split(',')))
    target_size_mb = float(input("Enter target minimum file size in MB (default 10): ") or 10)

    upscale_to_large_jpeg(input_file, output_file, scale, interpolation, bg_color, target_size_mb)
```
